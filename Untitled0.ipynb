```python
import os
from typing import TypedDict, Annotated, Sequence
from langgraph.graph import StateGraph, END
from langchain_groq import ChatGroq
from duckduckgo_search import DDGS
from supabase import create_client, Client

# ========================================================
# CONFIGURAÇÃO DE SOBERANIA (CHAVES)
# ========================================================
GROQ_API_KEY=gsk_pXWf2mYh6G8W6F5j7K9L0M1N2O3P4Q5R6S7T8U9V0W1X2Y3Z
SUPABASE_URL=https://jyfurrvkqrdkwtvtfzbw.supabase.co
SUPABASE_KEY=eyJhbGciOiJIUzI1NiIsInR5cCI6IkpXVCJ9.eyJpc3MiOiJzdXBhYmFzZSIsInZiI6Imp5ZnVycnZrcXJka3d0dnRmemJ3Iiwicm9sZSI6ImFub24iLCJpYXQiOjE3NTc5MTYzMjAsImV4cCI6MjA3MzQ5MjMyMH0.zuTYPgiy4PbsGdkG_rDX-YREhWcy225U2732Lq__Pno
MODELO_LOCAL=llama3.2:1b

# Inicialização dos Sistemas de Armas
llm = ChatGroq(temperature=0, model_name="llama-3.3-70b-versatile", groq_api_key=GROQ_API_KEY)
db: Client = create_client(SUPABASE_URL, SUPABASE_KEY)

# Estrutura de Dados do Sistema
class NexoState(TypedDict):
    missao: str        # Alvo (ex: Arbitragem de GPU)
    pesquisa: str      # Dados brutos colhidos
    analise: str       # Veredito da IA
    decisao: str       # LUCRO ou DESCARTAR
    log: str           # Status de execução

# ========================================================
# AGENTES (NÓS DO GRAFO)
# ========================================================

def caçador_node(state: NexoState):
    """ Varre a web em busca de erros de preço e oportunidades """
    print(f"🕵️ CAÇADOR: Iniciando varredura para: {state['missao']}")
    with DDGS() as ddgs:
        try:
            results = [r['body'] for r in ddgs.text(state['missao'], max_results=5)]
            return {"pesquisa": "\n".join(results), "log": "Busca concluída."}
        except Exception as e:
            return {"pesquisa": "Falha na busca", "log": f"Erro: {str(e)}"}

def critico_node(state: NexoState):
    """ Analisa o ROI e a probabilidade de lucro real """
    print("🧠 CRÍTICO: Analisando viabilidade econômica...")
    prompt = f"""
    COGNITIVO NEXO v12: Analise estes dados de mercado:
    {state['pesquisa']}
    Para a missão: {state['missao']}

    Responda EXATAMENTE neste formato:
    DECISAO: [LUCRO] ou [DESCARTAR]
    ROI_ESTIMADO: [%]
    JUSTIFICATIVA: [Curta e grossa]
    """
    res = llm.invoke(prompt).content
    decisao = "LUCRO" if "LUCRO" in res.upper() else "DESCARTAR"
    return {"analise": res, "decisao": decisao, "log": "Análise finalizada."}

def memoria_node(state: NexoState):
    """ Persistência absoluta no Supabase """
    print("💾 MEMÓRIA: Sincronizando com a Nuvem...")
    data = {
        "missao": state['missao'],
        "analise": state['analise'],
        "decisao": state['decisao'],
        "status": "PROCESSADO"
    }
    try:
        db.table("nexo_operacoes").insert(data).execute()
        return {"log": "Soberania de dados garantida no Supabase."}
    except Exception as e:
        print(f"❌ Erro de Banco: {e}")
        return {"log": "Falha na persistência."}

# ========================================================
# CONSTRUÇÃO DO MAESTRO (FLOW)
# ========================================================
workflow = StateGraph(NexoState)

workflow.add_node("caçador", caçador_node)
workflow.add_node("critico", critico_node)
workflow.add_node("memoria", memoria_node)

workflow.set_entry_point("caçador")
workflow.add_edge("caçador", "critico")
workflow.add_edge("critico", "memoria")
workflow.add_edge("memoria", END)

nexo_app = workflow.compile()

# ========================================================
# GATILHO DE OPERAÇÃO CONTÍNUA
# ========================================================
if __name__ == "__main__":
    missoes = [
        "iPhone 15 Pro Max preço abaixo de 5000 marketplace brasil",
        "RTX 4090 erro de preço estoque leilão",
        "Arbitragem de Bitcoin taxas entre Binance e Kraken hoje"
    ]

    for m in missoes:
        print(f"\n🚀 DISPARANDO MAESTRO PARA: {m}")
        nexo_app.invoke({"missao": m})
```
